### List of cases - Multiple variable

2D because it's easier to understand and visualize

Two 2D variables:
* Overlap = 1
* Overlap < 1

One 2D variable and one 1D variable:
* Overlap = 1
* Overlap < 1

Three 2D variables:
* Overlap = 1
* Overlap < 1

Two 2D variables and one 1D variable:
* Overlap = 1
* Overlap < 1

Each case is written to its own folder and rendered as a representative map.

The multi-variable examples now also accept per-variable overlap targets (`overlap=[...]`) and fixed overlap flags (`fixed_overlap=[...]`).


### Two 2D variables:

#### Overlap = 1

```python
ds, df = run_case(
    "overlap_eq1",
    num_obs=8,
    density=[8/9,8/9],
    num_vars=2,
    var_dims=2,
    overlap=1,
    fixed_overlap=False,
    seed=12345,
    base_dir="./tutorial3",
    title="Overlap = 1",
)
```

#### Overlap < 1

All but one measurements of Var1 are at the same sites of measurements for Var0

```python
ds, df = run_case(
    "overlap_le1",
    num_obs=8,
    density=[8/9,8/9],
    num_vars=2,
    var_dims=2,
    overlap=7/8,
    fixed_overlap=False,
    seed=12345,
    base_dir="./tutorial3",
    title="Overlap < 1",
)
```
